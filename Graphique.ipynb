## Interface Graphique

```python
# Version pour un placement automatique des widgets

from tkinter import * 

# Création d'une fenêtre
fenetre = Tk()

# Création d'un texte
mon_texte = Label(fenetre,text = "MON TEXTE AHAHAHAH")

# Affichage du texte dans la fenêtre - placement automatique
mon_texte.pack()

mon_texte1 = Label(fenetre,text = "2nd texte prout prout")
mon_texte1.pack()

# Boucle
fenetre.mainloop()

```

```python
# Version pour un placement un peu contrôlé des widgets

from tkinter import *
from tkinter import ttk

# Création d'une fenêtre
fenetre = Tk()

# Création d'un cadre dans la fenêtre et d'une grille
frm = ttk.Frame(fenetre)
frm.grid()

# Création d'un texte en vue d'un placement dans le cadre
objet=ttk.Label(frm, text = "Texte 1 qui est normal")
objet1=ttk.Label(frm, text = "Texte mais différent de l'autre")
objet2=ttk.Label(frm, text = "Texte cA 3 montre")

#Affichage du texte dans le cadre, selon les coordonnées précisés dans la grille
objet.grid(column=0, row=0)
objet1.grid(column=1, row=0)
objet2.grid(column=2, row=1)
texte_user.grid(column=0, row=2)
saisie = ttk.Entry(frm, bd = 5, bg="green")

fenetre.mainloop()
```
