```python
def moyenne_ponderee():

    liste_note = [14, 7, 18, 3, 11, 20, 9, 15, 0, 12, 6, 17, 4, 19, 10, 8, 13, 2, 16, 5, 12]
    coefNoteUE1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
    coefNoteUE2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
    coefNoteUE3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

    total_point = 0 # Variable local pour la suite (Total Point) (Notes x Coef)
    total_coef = 0 # Variable local pour la suite (Total Coef)
    for i in range(21): # I va parcourir chaque valeur de 0 à 21
        total_point += liste_note[i] * coefNoteUE1[i] # Multiplie les notes avec coefs pour avoir le total_point toujours en prenant compte [i]
        total_coef += coefNoteUE1[i] # Fait la somme de tout les coefs dans total_coef toujours en prenant compte [i]
    moyenneUE1 = total_point / total_coef # Calcule la moyenne
    print("Voici votre moyenne dans l'UE1 = ", moyenneUE1) # Print le résultat

    total_point = 0
    total_coef = 0
    for i in range(21):
        total_point += liste_note[i] * coefNoteUE2[i]
        total_coef += coefNoteUE2[i]
    moyenneUE2 = total_point / total_coef
    print("Voici votre moyenne dans l'UE2 = ", moyenneUE2)

    total_point = 0
    total_coef = 0
    for i in range(21):
        total_point += liste_note[i] * coefNoteUE3[i]
        total_coef += coefNoteUE3[i]
    moyenneUE3 = total_point / total_coef
    print("Voici votre moyenne dans l'UE3 = ", moyenneUE3)
    
moyenne_ponderee()
```

```python
import numpy as np

# création de la liste de toutes les matières du semestre 1
matiere = [f"R{i}" for i in range(101, 116)]+[f"SAÉ{i}" for i in range(11,17)]

# lites des coefficients des trois UE
list_coef1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
list_coef2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
list_coef3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

# boucle pour récupérer la liste des notes (version "comprehension list")
# list_notes = [float(input(f"Quelle est la note pour {element} ?")) for element in matiere]

list_notes = [float(input(f"Quelle est la note pour {element} ?")) for element in matiere]

"""
dico_notes = {cle : valeur for cle, valeur in zip(matiere, list_notes)}
dico_coef1 = {cle : valeur for cle, valeur in zip (matiere, list_coef1)}
dico_coef2 = {cle : valeur for cle, valeur in zip (matiere, list_coef2)}
dico_coef3 = {cle : valeur for cle, valeur in zip (matiere, list_coef3)}
"""

def dico(liste):
    dico = {cle : valeur for cle, valeur in zip(matiere, liste)}
    return dico

dico_notes = dico(list_notes)
dico_coef1 = dico(list_coef1)
dico_coef2 = dico(list_coef2)
dico_coef3 = dico(list_coef3)
```

```python
# Calcul de la moyenne pondérée "à la main" comme dans la version 1

total_coef = 0
somme = 0
for cle in dico_coef1.keys(): # pour chaque clé, donc pour chaque matière
    total_coef += dico_coef1[cle] # cumul des coefficients (valeur récupérées dans le dict des coeffs)
    somme += dico_coef1[cle] * dico_notes[cle] # cumul des produits note*coeff avec valeurs récup dans dict
moy = round((somme/total_coef), 3) # calcul de la moyenne avec 3 chiffres après la virgule
print(f"La moyenne de l'UE1 vaut {moy}")
```
